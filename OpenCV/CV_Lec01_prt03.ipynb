# Topic: GrayScale Coversion 

```python
# import lib
import cv2 as cv
from cv2 import cvtColor
```

```python
img = cv.imread('Resources/beautifull_bird.jpg')
img = cv.resize(img , (800 , 600))

# Conversion
gray_scale = cvtColor(img , cv.COLOR_BGR2GRAY)

# Display Code
cv.imshow('WithoutGrayScale / AlreadyExistImage' , img)
cv.imshow('GrayScaleImage' , gray_scale)

# Delay
cv.waitKey(0)
```

```python
# Close All Windows
cv.destroyAllWindows()
```

```python
import cv2 as cv
from cv2 import cvtColor

# read img
r_img = cv.imread("Resources/beautifull_bird.jpg")

# resize img
rsize_img = cv.resize(r_img , (800 , 700))

# conversion 
gray_scale = cvtColor(rsize_img , cv.COLOR_BGR2GRAY)

# show img 

show_without_rsize_img = cv.imshow('Colorfull Without (Resize)' , r_img)

show_colorfull_img = cv.imshow('Colorfull With (Resize)' , rsize_img)
print(show_colorfull_img)

show_color_less_img = cv.imshow('Colorless With(Resize)' , gray_scale)
print(show_color_less_img)

# delay img time
dly_img = cv.waitKey(0)
print(dly_img)
```

```python
# close all windows
c_windows = cv.destroyAllWindows()
print(c_windows)
```
