```python
import cv2 as cv

# read image
brain_img = cv.imread("Resources/brain_power.jpg")
# resize image / set image size
resize_img = cv.resize(brain_img , (400 , 400))

# show image
show_resize_img = cv.imshow("Three Rotations Are Brain(Resize)" , resize_img)
print(show_resize_img)

show_img = cv.imshow("Three Rotations Are Brain" , brain_img)
print(show_img)

# delay image
delay_img = cv.waitKey(0)
print(delay_img)
```

```python
import cv2 as cv

# read img 
rd_img = cv.imread("Resources/brain_power.jpg")

# resize image
rsize_img = cv.resize(rd_img, (600 , 800))

# show img
show_img = cv.imshow("3 types of Brain Rotation" , rd_img)
print(show_img)

show_rsize_img = cv.imshow("3 types of Brain Rotation(Resize Image)" , rsize_img)
print(show_rsize_img)

# delay image
dly_img = cv.waitKey(0)
print(dly_img)
```
